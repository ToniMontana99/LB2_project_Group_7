```python
import pandas as pd
```

```python
train = pd.read_csv("training.tsv", sep="\t", header=None, names=["accession", "label"])
train["accession"] = train["accession"].str.strip()
neg = pd.read_csv("negative_data.tsv", sep="\t")
pos = pd.read_csv("positive_data.tsv", sep="\t")
uni = pd.concat([pos, neg])
```

```python
out = train.merge(uni, on="accession", how="left")
out.to_csv("training_annotated.tsv", sep="\t", index=False)
```
